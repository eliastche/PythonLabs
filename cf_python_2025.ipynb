**Exercice 1** \
Write a Python program that:
* Asks the user to input the daily return of stocks (as percentage)
* Prints:
  * "Profit" if the return is positive
  * "Loss" if the return is negative
  * "No change" if its exactly zero
* With the additional conditions:
  * If the return is greater than 2%, print "Strong gain" instead
  * If the return is less than -2%, print "Strong loss" instead

```python
def dailyreturns():
  returnstock = float(input("Input the daily return of a stock in precentage: "))
  returnresult = "No change"

  if returnstock > 0:
    returnresult = "Profit"
  elif returnstock < 0:
    returnresult = "Loss"

  return print("The stock has", returnresult)

dailyreturns()

```

```python
#With additional constraints

def dailyreturns():
  returnstock = float(input("Input the daily return of a stock in precentage: "))
  returnresult = "No change"

  if 0 < returnstock <= 2:
    returnresult = "Profit"
  elif -2 <=returnstock < 0:
    returnresult = "Loss"
  elif 2 < returnstock:
    returnresult = "Strong gain"
  elif returnstock < -2:
    returnresult = "Strong loss"

  return print("The stock has", returnresult)

dailyreturns()
```

**Excercise 2**\
Write a python script that simulates the growth of an investment over several years
* Start with an initital capital of 1000
* For each year from 1 to 5, assume a 5% annual return
* Use a for loop to compute and print the capital at the end of each year liks:
  * Year 1: 1050€
  * Year 2: 1102.5€
  * ...
* After the loop print the total gain (final - initial)

```python
def growthinvestment(S0):
  r = 0.05
  years = [1,2,3,4,5]
  S = S0

  for y in years:
    S = S*(1+r)
    print(f"Year {y}: {S}€")

  gain = S-S0

  print(f"The total gain is: {gain}€")

growthinvestment(1000)
```
