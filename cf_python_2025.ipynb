**Exercice 1** \
Write a Python program that:
* Asks the user to input the daily return of stocks (as percentage)
* Prints:
  * "Profit" if the return is positive
  * "Loss" if the return is negative
  * "No change" if its exactly zero
* With the additional conditions:
  * If the return is greater than 2%, print "Strong gain" instead
  * If the return is less than -2%, print "Strong loss" instead

```python
def dailyreturns():
  returnstock = float(input("Input the daily return of a stock in precentage: "))
  returnresult = "No change"

  if returnstock > 0:
    returnresult = "Profit"
  elif returnstock < 0:
    returnresult = "Loss"

  return print("The stock has", returnresult)

dailyreturns()

```

```python
#With additional constraints

def dailyreturns():
  returnstock = float(input("Input the daily return of a stock in precentage: "))
  returnresult = "No change"

  if 0 < returnstock <= 2:
    returnresult = "Profit"
  elif -2 <=returnstock < 0:
    returnresult = "Loss"
  elif 2 < returnstock:
    returnresult = "Strong gain"
  elif returnstock < -2:
    returnresult = "Strong loss"

  return print("The stock has", returnresult)

dailyreturns()
```

**Excercise 2**\
Write a python script that simulates the growth of an investment over several years
* Start with an initital capital of 1000
* For each year from 1 to 5, assume a 5% annual return
* Use a for loop to compute and print the capital at the end of each year liks:
  * Year 1: 1050€
  * Year 2: 1102.5€
  * ...
* After the loop print the total gain (final - initial)

```python
def growthinvestment(S0):
  r = 0.05
  years = [1,2,3,4,5]
  S = S0

  for y in years:
    S = S*(1+r)
    print(f"Year {y}: {S}€")

  gain = S-S0

  print(f"The total gain is: {gain}€")

growthinvestment(1000)
```

##Part 2: Data analysis and manipulation
**1. Load some stocks information into a pandas DataFrame**
* Using the yFinance module, use this code o download stock information from Apple, Microsoft and the S&P 500 index

```python
!pip install yfinance
import yfinance as yf
tickers = ['AAPL', 'MSFT', '^GSPC']
df = yf.download(tickers, start='2018-01-01')

aapl_stock = df.xs('AAPL', level = 1, axis = 1)
msft_stock = df.xs('MSFT', level = 1, axis = 1)
gspc_stock = df.xs('^GSPC', level = 1, axis = 1)

aapl_stock.columns
```

**a. Compute the returns of the 'Close' column and inspect the new dataframe**

```python
aapl_stock['Return'] = aapl_stock['Close'].pct_change()
msft_stock['Return'] = msft_stock['Close'].pct_change()
gspc_stock['Return'] = gspc_stock['Close'].pct_change()

aapl_stock['Return'].head()
```

**b. Compute the correlation of the returns of AAPL and MSFT with S&P 500**

```python
!pip install pandas
import pandas as pd

merge_aapl_gspc = pd.merge(left=aapl_stock['Return'], right=gspc_stock['Return'], left_index=True, right_index=True, suffixes=('_sufx1', '_sufx2'))
merge_msft_gspc = pd.merge(left=msft_stock['Return'], right=gspc_stock['Return'], left_index=True, right_index=True, suffixes=('_sufx1', '_sufx2'))

merge_aapl_gspc = merge_aapl_gspc.dropna()
merge_msft_gspc = merge_msft_gspc.dropna()

#The correlations
aapl_gspc_corr = merge_aapl_gspc[['Return_sufx1', 'Return_sufx2']].corr()
msft_gspc_corr = merge_msft_gspc[['Return_sufx1', 'Return_sufx2']].corr()

print("Correlation between Apple and the S&P 500", aapl_gspc_corr)
print("Correlation between Microsoft and the S&P 500", msft_gspc_corr)
```

**c. Compute the cumulative returns by calculating the cumulative product over the returns. Store the results in a new dataframe, and plot it by calling the pandas plot function**

```python
import numpy as np

aapl_stock['cumu_return'] = np.cumprod(aapl_stock['Return']+1)
msft_stock['cumu_return'] = np.cumprod(msft_stock['Return']+1)
gspc_stock['cumu_return'] = np.cumprod(gspc_stock['Return']+1)

gspc_stock['cumu_return'].head()
```

**d. Compute the annual returns assuming that there are 253 trading days per year**

```python
aapl_stock['Annual Return'] = np.mean(aapl_stock['Return'])*252
msft_stock['Annual Return'] = np.mean(msft_stock['Return'])*252
gspc_stock['Annual Return'] = np.mean(gspc_stock['Return'])*252

aapl_stock.head()
```
